# 3. Single-Qubit Pauli Gates

## Challenge 2/3: Binary numbers

```python
from qiskit import QuantumCircuit, ClassicalRegister, transpile
from qiskit import *
# from qiskit import Aer, *
from qiskit_aer import Aer
from qiskit.visualization import plot_state_qsphere
%matplotlib inline
```

```python
circuit = QuantumCircuit(4)
circuit.x([0, 2 , 3])
circuit.draw(output='mpl')
```

```python
simulator = Aer.get_backend('statevector_simulator')
compiled_circuit = transpile(circuit, backend=simulator)
state_vector_result = simulator.run(compiled_circuit).result()
state_vector = state_vector_result.get_statevector()
plot_state_qsphere(state_vector)
```
