```python
from qiskit import QuantumCircuit, ClassicalRegister, transpile
from qiskit import *
# from qiskit import Aer, *
from qiskit_aer import Aer
from qiskit.visualization import plot_histogram, array_to_latex
%matplotlib inline
```

## Base Circuit: Linear Entanglement (4-qubit GHZ)
This code builds a 4-qubit entanglement circuit using a linear CNOT chain:
- Apply `H` on qubit 0 to create superposition.
- Apply `CX` gates `0->1`, `1->2`, `2->3` to propagate entanglement.

This prepares a GHZ-type state (also called a cat state):
$$(|0000\rangle + |1111\rangle)/\sqrt{2}$$

```python
circuit = QuantumCircuit(4)

circuit.h(0)
circuit.cx(0, 1)
circuit.cx(1, 2)
circuit.cx(2, 3)

circuit.draw(output='mpl')
```

```python
simulator = Aer.get_backend('statevector_simulator')
compiled_circuit = transpile(circuit, backend=simulator)
result = simulator.run(compiled_circuit).result()
state_vector = result.get_statevector()
array_to_latex(state_vector, prefix="\\text{StateVector = }\n")
```

```python
circuit.measure_all()
qasm_simulator = Aer.get_backend('qasm_simulator')
compiled_circuit = transpile(circuit, backend=qasm_simulator)
result = qasm_simulator.run(compiled_circuit, shots=1024).result()
plot_histogram(result.get_counts())
```
